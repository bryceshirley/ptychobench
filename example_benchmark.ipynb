```python
# Even if the file is inside src/ptychobench/solver/solver.py
from ptychobench.grid import SimulationGrid
from ptychobench.operators import (
    ExactOperator,
    ParaxialOperator,
    FeitFleckOperator,
    LinDudaOperator,
)
from ptychobench.samples import zig_balls, zigs, straight_balls, straight
from ptychobench.solver import run_benchmark
from ptychobench.utils import generate_results
```

```python
grid = SimulationGrid(divergence_angle=20.0)
sample_params = {"modulus": 0.75, "phase": 0.0075}
```

```python
operators = [ExactOperator, ParaxialOperator, FeitFleckOperator, LinDudaOperator]
```

```python
result = run_benchmark(
    grid=grid, sample=zig_balls, operators=operators, sample_params=sample_params
)
generate_results(grid, result)
```

```python
result = run_benchmark(
    grid=grid, sample=zigs, operators=operators, sample_params=sample_params
)
generate_results(grid, result)
```

```python
result = run_benchmark(
    grid=grid, sample=straight_balls, operators=operators, sample_params=sample_params
)
generate_results(grid, result)
```

```python
result = run_benchmark(
    grid=grid, sample=straight, operators=operators, sample_params=sample_params
)
generate_results(grid, result)
```
